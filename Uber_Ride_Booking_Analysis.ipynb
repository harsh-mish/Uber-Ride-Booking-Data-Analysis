```python
# 1. Date and Time
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
time_parsed = pd.to_datetime(df["Time"].astype(str), format="%H:%M:%S", errors="coerce")
df["Time"] = time_parsed.dt.strftime("%H:%M:%S")
df["Hour"] = time_parsed.dt.hour
```

```python
df
```

```python
df["Time"].head(10)
```

```python
df["Day_of_Week"] = df["Date"].dt.day_name()
df["Day_Number"] = df["Date"].dt.day
df["Month"] = df["Date"].dt.month_name()
df["Month_Number"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year
```

```python
df
```

```python
# 3. Time period
def get_time_period(hour):
    if pd.isna(hour):
        return "Unknown"
    if 5 <= hour < 9:
        return "Early Morning"
    elif 9 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df["Time_Period"] = df["Hour"].apply(get_time_period)
```

```python
# 4. Standardized cancellation/status category
def get_category(status):
    status = str(status).strip().lower()

    if "customer" in status and "cancel" in status:
        return "Customer Cancelled"
    elif "driver" in status and "cancel" in status:
        return "Driver Cancelled"
    elif "incomplete" in status:
        return "Incomplete"
    elif "no driver" in status:
        return "No Driver Found"
    elif "complete" in status:
        return "Completed"
    else:
        return "Other"

df["Cancellation_Category"] = df["Booking Status"].apply(get_category)
```

```python
df
```

```python
# 5. Data-quality checks
print("Dataset shape:", df.shape)
print("Duplicate Booking IDs:", df["Booking ID"].duplicated().sum())
print("Duplicate complete rows:", df.duplicated().sum())
print("Zero Booking Value:", (df["Booking Value"] == 0).sum())
print("Zero Ride Distance:", (df["Ride Distance"] == 0).sum())

unknown_payment = (
    df["Payment Method"].astype(str).str.strip().str.lower().eq("unknown").sum()
)
print("Unknown Payment Method:", unknown_payment)
```

```python
# 6. Important missing-value check
important_columns = [
    "Date", "Time", "Booking ID", "Booking Status", "Customer ID",
    "Vehicle Type", "Pickup Location", "Drop Location",
    "Booking Value", "Ride Distance", "Payment Method"
]

print("\nMissing values in important columns:")
display(df[important_columns].isna().sum().to_frame("Missing_Count"))
```

## UniVariate Analysis

```python
numeric_cols = [
    "Booking Value",
    "Ride Distance",
    "Avg VTAT",
    "Avg CTAT",
    "Driver Ratings",
    "Customer Rating",
    "Hour"
]

corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")
plt.show()
```

```python
pairplot_cols = [
    "Booking Value",
    "Ride Distance",
    "Driver Ratings",
    "Customer Rating"
]

pairplot_df = df[pairplot_cols].dropna().sample(
    5000,
    random_state=42
)

sns.pairplot(
    pairplot_df,
    diag_kind="kde"
)

plt.show()
```
